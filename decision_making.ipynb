```python
'''
Scenario: A school wants to classify roll numbers as even or odd for seating arrangements.
Detailed Description: Write a Python program to check whether a given number is even or odd using if-else statements.
Input Format: A single integer N. Output Format: Print 'Even' or 'Odd'.
Constraints: 1 <= N <= 10^6
Sample Input: 7 Sample Output: Odd
Explanation: Since 7 is not divisible by 2, the output is Odd.
'''


n = int(input())
if 1 <= n <= 10**6:
    if n % 2 == 0:
        print("Even")
    else:
        print("Odd")
else:
    print("Number out of range")
```
