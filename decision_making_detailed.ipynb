```python
'''
 Internet Speed Category
Scenario: An internet service provider wants to classify customer internet speed.
Detailed Description: Read the internet speed in Mbps and print whether the speed is Slow, Moderate, or Fast.
Rules:
• Below 10 → Slow
• 10 to 49 → Moderate
• 50 and above → Fast
Input Format: A single integer S representing internet speed.
Output Format: Print the speed category.
Constraints: 0 <= S <= 1000
Test Cases:
Sample Input
55
8
Sample Output
Fast
Slow
25
Moderate
'''

s = int(input())
if 0 <= s <= 1000:
  if s < 10:
    print("slow")
  elif 10 <= s <= 49:
    print("Moderate")

  else:
    print("Fast")
else:
  print("out of range")
```

```python
'''
Mobile Battery Alert
Scenario: A smartphone displays alerts depending on battery level.
Detailed Description: Read battery percentage and display the battery status.
Rules:
• 0–10 → Critical
• 11–30 → Low Battery
• Above 30 → Battery OK
Input Format: A single integer B representing battery percentage.
Output Format: Print battery status.
Constraints: 0 <= B <= 100
Test Cases:
Sample Input
8
20
75
Sample Output
Critical
Low Battery
Battery OK
'''
B = int(input())
if 0 <= B <= 100:
  if 0 <= B <= 10:
    print("Critical")
  elif 11 <= B <= 30:
    print("Low Battery")
  else:
    print("Battery OK")
else:
  print("out of range")
```

```python
'''
 ATM Withdrawal Check
Scenario: A bank ATM accepts withdrawals only in multiples of 100.
Detailed Description: Read withdrawal amount and check if the amount is valid.
Rules:
• Divisible by 100 → Transaction Allowed
• Otherwise → Invalid Amount
Input Format: A single integer A representing withdrawal amount.
Output Format: Print transaction status.
Constraints: 100 <= A <= 100000
'''

A = int(input())
if 100 <= A <= 100000:
  if A % 100 == 0:
    print("Transaction Allowed")
  else:
    print("Invalid Amount")
else:
  print("out of range")

```

```python
'''
4. Temperature Warning System
Scenario: A weather station monitors outside temperature.
Invalid Amount
Detailed Description: Read temperature and display weather condition.
Rules:
• Below 15 → Cold
• 15 to 30 → Normal
• Above 30 → Hot
Input Format: A single integer T.
Output Format: Print temperature category.
Constraints: -50 <= T <= 60
'''
T = int(input())
if  -50 <= T <= 60:
  if T < 15:
    print("Cold")
  elif 15 <= T <= 30:
    print("Normal")
  else:
    print("Hot")
else:
  print("out of range")
```
