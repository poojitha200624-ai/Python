```python
'''
1. Tax Slab Identifier - Annual Income Advisory
Scenario:
A finance app helps users understand which tax category their annual income falls into before calculating detailed tax
later.
Detailed Description:
Write a Python program that reads an annual income and prints the correct tax slab name. Use if, elif, and else
decision-making statements only.
Input Format:
A single integer income representing annual income in rupees.
Output Format:
Print No Tax Slab, Basic Slab, Standard Slab, Higher Slab, or Premium Slab.
Constraints:
0 <= income <= 10^9
Sample Input:
850000
Sample Output:
Higher Slab
Explanation:
850000 is greater than 750000 and less than or equal to 1000000, so it belongs to the Higher Slab.
'''

income = int(input())
if 0 <= income <= 10**9:
  if income <= 250000:
    print("No tax slab")
  elif 250000 <= income <= 500000:
    print("Basic slab")
  elif 500000 <= income <= 750000:
    print("Standard slab")
  elif 750000 <= income <= 1000000:
    print("Higher slab")
  elif income > 1000000:
    print("premium slab")
  else:
    print("Invalid input")
else:
  print("out of range")
```

```python
'''
2. Scholarship Eligibility - Merit and Income Filter
Scenario:
A college scholarship portal approves students based on both academic performance and family income.
Detailed Description:
Read a student percentage and annual family income. A student gets Full Scholarship if percentage is at least 90 and
income is at most 200000. They get Half Scholarship if percentage is at least 80 and income is at most 400000.
Otherwise, they are Not Eligible.
Input Format:
Two values: percentage P as an integer and income I as an integer.
Output Format:
Print Full Scholarship, Half Scholarship, or Not Eligible.
Constraints:
0 <= P <= 100; 0 <= I <= 10^7
Sample Input:
86 350000
Sample Output:
Half Scholarship
Explanation:
The percentage is at least 80 and income is at most 400000, so the student receives Half Scholarship.
'''
P, I = map(int, input().split())

if 0 <= P <= 100 and 0 <= I <= 10**7:
    if P >= 90 and I <= 200000:
        print("Full Scholarship")
    elif P >= 80 and I <= 400000:
        print("Half Scholarship")
    else:
        print("Not Eligible")
else:
    print("out of range")
```

```python
'''
3. Smart Parking Fee - Vehicle and Duration Based
Scenario:
A smart parking system charges different rates based on vehicle type and number of parked hours.
Detailed Description:
Read vehicle type and hours. For bike, charge 20 per hour for first 3 hours and 10 for each extra hour. For car, charge 50
per hour for first 3 hours and 30 for each extra hour. For any other vehicle type, print Invalid Vehicle.
Input Format:
A string vehicle and an integer hours. Vehicle will be entered in lowercase.
Output Format:
Print the total parking fee as an integer, or Invalid Vehicle.
Constraints:
vehicle length <= 20; 1 <= hours <= 24
Sample Input:
car 5
Sample Output:
210
Explanation:
For car: first 3 hours cost 3 * 50 = 150 and extra 2 hours cost 2 * 30 = 60. Total = 210.
'''
vehicle, hours = input().split()
hours = int(hours)

if vehicle == "bike":
    if hours <= 3:
        fee = hours * 20
    else:
        fee = (3 * 20) + ((hours - 3) * 10)
    print(fee)

elif vehicle == "car":
    if hours <= 3:
        fee = hours * 50
    else:
        fee = (3 * 50) + ((hours - 3) * 30)
    print(fee)

else:
    print("Invalid Vehicle")
```

```python
'''
4. Bank Loan Decision - Credit Score and Salary Check
Scenario:
A bank wants to quickly classify loan applications before detailed manual verification.
Detailed Description:
Read credit score and monthly salary. If credit score is at least 750 and salary is at least 50000, print Approved. If credit
score is at least 650 and salary is at least 30000, print Review Required. Otherwise, print Rejected.
Input Format:
Two integers C and S representing credit score and monthly salary.
Output Format:
Print Approved, Review Required, or Rejected.
Constraints:
300 <= C <= 900; 0 <= S <= 10^7
Sample Input:
680 42000
Sample Output:
Review Required
Explanation:
The applicant does not satisfy the Approved condition, but satisfies the Review Required condition.
'''

C, S = map(int, input().split())
if 300 <= C <= 900 and 0 <= S <= 10**7:
  if C >= 750 and S >= 50000:
    print("Approved")
  elif C >= 650 and S >= 30000:
    print("Review Required")
  else:
    print("Rejected")
else:
  print("Invalid")
```

```python
'''
5. Triangle Type Analyzer - Construction Material Check
Scenario:
An engineering tool checks whether three rods can form a triangle and identifies the triangle type.
Detailed Description:
Read three side lengths. First check whether they form a valid triangle. If invalid, print Invalid Triangle. If valid, print
Equilateral, Isosceles, or Scalene.
Input Format:
Three integers A, B, and C representing side lengths.
Output Format:
Print Invalid Triangle, Equilateral, Isosceles, or Scalene.
Constraints:
1 <= A, B, C <= 10^6
Sample Input:
6 6 10
Sample Output:
Isosceles
Explanation:
6 + 6 > 10, so the triangle is valid. Two sides are equal, so it is an Isosceles triangle.
'''
A, B, C = map(int,input().split())
if 1 <= A <= 10**6 and 1 <= B <= 10**6 and 1 <= C <= 10**6 :
   if A + B > C and A + C > B and B + C > A:
    if A == B == C:
      print("Equilateral")
    elif A == B or A == C or B == C:
      print("isoceles")
    else:
      print("Scalene")
   else:
    print("Invalid Triangle")
else:
  print("out of range")



```

```python
'''
6. Exam Result Status - Marks and Attendance Rule
Scenario:
A school declares final result using both exam marks and attendance percentage.
Detailed Description:
Read marks and attendance. If attendance is below 75, print Detained. If attendance is at least 75 and marks are at least
35, print Pass. Otherwise, print Fail.
Input Format:
Two integers M and A representing marks and attendance percentage.
Output Format:
Print Detained, Pass, or Fail.
Constraints:
0 <= M <= 100; 0 <= A <= 100
Sample Input:
72 68
Sample Output:
Detained
Explanation:
Even though the mark is good, attendance is below 75, so the student is Detained.
'''
M, A = map(int,input().split())
if 0 <= M <= 100 and 0 <= A <= 100:
  if A < 75:
    print("Detained")
  elif A >= 75 and M >= 35:
    print("Pass")
  else:
    print("Fail")
else:
  print("Invalid")
```

```python
'''
7. Delivery Charge Calculator - Distance and Order Amount
Scenario:
A food delivery app calculates delivery charges based on order amount and delivery distance.
Detailed Description:
Read order amount and distance in kilometers. If amount is at least 1000, delivery is free. Otherwise, distance up to 5 km
costs 40, distance up to 10 km costs 70, and distance above 10 km costs 120.
Input Format:
Two integers amount and distance.
Output Format:
Print the delivery charge as an integer.
Constraints:
0 <= amount <= 10^7; 1 <= distance <= 100
Sample Input:
650 12
Sample Output:
120
Explanation:
The order amount is below 1000 and distance is above 10 km, so the delivery charge is 120.
'''
amount, distance = map(int, input().split())

if 0 <= amount <= 10**7 and 1 <= distance <= 100:
    if amount >= 1000:
        print(0)
    elif distance <= 5:
        print(40)
    elif distance <= 10:
        print(70)
    else:
        print(120)
else:
    print("Invalid")
```

```python
'''
Movie Ticket Pricing - Age, Day, and Seat Type
Scenario:
A cinema booking system changes ticket price based on age group, weekday/weekend, and seat type.
Detailed Description:
Read age, day type, and seat type. Base price is 120 for normal seat and 200 for premium seat. On weekend add 50. If
age is below 12, give 50% discount on the final price. If seat type is invalid, print Invalid Seat.
Input Format:
An integer age, followed by strings day and seat. day is weekday or weekend. seat is normal or premium.
Output Format:
Print the final ticket price as an integer, or Invalid Seat.
Constraints:
0 <= age <= 120; day and seat are lowercase strings
Sample Input:
10 weekend premium
Sample Output:
125
Explanation:
Premium price is 200. Weekend adds 50, making 250. Child discount gives 50%, so final price is 125.
'''
day, seat=input().split()
age = int(input())
if 0 <= age <= 120 and day.islower() and seat.islower():
  if seat == "normal":
    price = 120
  elif seat == "premium":
    price = 200
  else:
    print("invalid seat")
  if day == "weekend":
    price = price + 50
  if age < 12:
    price = price * 0.5
    print(price)
  else:
    print(price)
else:
  print("Invalid")
```

```python
'''
9. Water Tank Alert - Level and Motor Status
Scenario:
An automatic water tank monitor sends alerts based on current water level and motor status.
Detailed Description:
Read water level percentage and motor status. If level is below 20 and motor is off, print Start Motor. If level is above 90
and motor is on, print Stop Motor. If level is between 20 and 90 inclusive, print Normal. Otherwise, print Check Sensor.
Input Format:
An integer level and a string status. status is on or off.
Output Format:
Print Start Motor, Stop Motor, Normal, or Check Sensor.
Constraints:
0 <= level <= 100; status will be on or off
Sample Input:
95 on
Sample Output:
Stop Motor
Explanation:
The tank level is above 90 and the motor is running, so the motor should be stopped.
'''
level = int(input())
status = input()
if 0 <= level <= 100:
  if level < 20 and status == "off":
    print("Start Motor")
  elif level > 90 and status == "on":
    print("Stop motor")
  elif  20 <= level <= 90 and status == "on":
    print("Normal")
  else:
    print("Check Sensor")
else:
  print("Invalid")
```

```python
'''
10. Password Strength Checker - Rule-Based Security
Scenario:
A login system wants to classify passwords using simple measurable rules before allowing registration.
Detailed Description:
Read password length, number of digits, and number of special characters. Print Strong if length is at least 8, digits are at
least 2, and special characters are at least 1. Print Medium if length is at least 6 and digits are at least 1. Otherwise, print
Weak.
Input Format:
Three integers L, D, and S representing length, digit count, and special character count.
Output Format:
Print Strong, Medium, or Weak.
Constraints:
0 <= L <= 100; 0 <= D, S <= L
Sample Input:
9 2 1
Sample Output:
Strong
Explanation:
All Strong conditions are satisfied: length is at least 8, digits are at least 2, and there is at least one special character.
'''
L, D, S = map(int, input().split())
if 0 <= L <= 100 and 0 <= D <= L and 0 <= S <= L:
  if L >= 8 and D >= 2 and S >= 1:
    print("Strong")
  elif L >= 6 and D >= 1:
    print("Medium")
  else:
    print("Weak")
else:
  print("Invalid")
```

```python
'''
11. Electricity Bill with Surcharge - Usage and Payment Delay
Scenario:
An electricity board calculates bill amount using slabs and adds a surcharge for late payment.
Detailed Description:
Read units consumed and delay days. First calculate bill: first 100 units at 2 per unit, next 100 units at 4 per unit,
remaining units at 6 per unit. If delay days are more than 10, add 5% surcharge. Test cases will make the final amount an
integer.
Input Format:
Two integers U and D representing units and delay days.
Output Format:
Print the final bill amount as an integer.
Constraints:
0 <= U <= 10000; 0 <= D <= 365
Sample Input:
250 15
Sample Output:
945
Explanation:
Bill = 100*2 + 100*4 + 50*6 = 900. Delay is more than 10 days, so 5% surcharge is added. Final = 945.
'''
U, D = map(int, input().split())

if 0 <= U <= 10000 and 0 <= D <= 365:

    if U <= 100:
        bill = U * 2
    elif U <= 200:
        bill = (100 * 2) + ((U - 100) * 4)
    else:
        bill = (100 * 2) + (100 * 4) + ((U - 200) * 6)

    if D > 10:
        bill = bill + (bill * 0.05)

    print(int(bill))

else:
    print("Invalid")
```

```python
'''
12. Employee Bonus Category - Experience and Rating
Scenario:
A company awards bonuses using both work experience and yearly performance rating.
Detailed Description:
Read years of experience and rating. If experience is at least 5 and rating is at least 4.5, print 20% Bonus. If experience is
at least 3 and rating is at least 4.0, print 10% Bonus. If rating is below 3.0, print No Bonus. Otherwise, print 5% Bonus.
Input Format:
An integer experience and a float rating.
Output Format:
Print 20% Bonus, 10% Bonus, 5% Bonus, or No Bonus.
Constraints:
0 <= experience <= 50; 0.0 <= rating <= 5.0
Sample Input:
4 4.2
Sample Output:
10% Bonus
Explanation:
Experience is at least 3 and rating is at least 4.0, so the employee gets 10% Bonus.
'''
experience, rating = input().split()
experience = int(experience)
rating = float(rating)
if 0 <= experience <= 50 and 0.0 <= rating <= 5.0:
    if experience >= 5 and rating >= 4.5:
        print("20% Bonus")
    elif experience >= 3 and rating >= 4.0:
        print("10% Bonus")
    elif rating < 3.0:
        print("No Bonus")
    else:
        print("5% Bonus")
else:
    print("Invalid")
```

```python
'''
13. ATM Withdrawal Validator - Balance, Amount, and PIN
Scenario:
An ATM machine must validate PIN, withdrawal amount, and available balance before dispensing cash.
Detailed Description:
Read balance, withdrawal amount, and PIN status. PIN status is valid or invalid. If PIN is invalid, print Invalid PIN. If
amount is not a multiple of 100, print Invalid Amount. If amount is greater than balance, print Insufficient Balance.
Otherwise, print Transaction Successful.
Input Format:
Two integers balance and amount, followed by string pinStatus.
Output Format:
Print Invalid PIN, Invalid Amount, Insufficient Balance, or Transaction Successful.
Constraints:
0 <= balance <= 10^7; 0 <= amount <= 10^7; pinStatus is valid or invalid
Sample Input:
5000 2300 valid
Sample Output:
Invalid Amount
Explanation:
The PIN is valid, but 2300 is not a multiple of 100, so the withdrawal amount is invalid.
'''
balance, amount, pinStatus = input().split()
balance = int(balance)
amount = int(amount)
if 0 <= balance <= 10**7 and 0 <= amount <= 10**7:
    if pinStatus == "invalid":
        print("Invalid PIN")
    elif amount % 100 != 0:
        print("Invalid Amount")
    elif amount > balance:
        print("Insufficient Balance")
    else:
        print("Transaction Successful")
else:
    print("Invalid")
```

```python
'''
14. Mobile Recharge Offer - Plan and User Type
Scenario:
A telecom app gives recharge benefits based on selected plan amount and user category.
Detailed Description:
Read recharge amount and user type. If user type is new and amount is at least 299, print Extra 2GB. If user type is
existing and amount is at least 499, print Cashback 50. If amount is below 100, print Invalid Recharge. Otherwise, print
Standard Benefits.
Input Format:
An integer amount and a string userType. userType is new or existing.
Output Format:
Print Extra 2GB, Cashback 50, Invalid Recharge, or Standard Benefits.
Constraints:
0 <= amount <= 10000; userType is lowercase
Sample Input:
499 existing
Sample Output:
Cashback 50
Explanation:
The user is existing and recharge amount is at least 499, so Cashback 50 is applied.
'''
amount, userType = input().split()
amount = int(amount)
if 0 <= amount <= 10000:
    if userType == "new" and amount >= 299:
        print("Extra 2GB")
    elif userType == "existing" and amount >= 499:
        print("Cashback 50")
    elif amount < 100:
        print("Invalid Recharge")
    else:
        print("Standard Benefits")
else:
    print("Invalid")
```

```python
''' 15. Online Exam Proctor Alert - Tab Switches and Camera Status
Scenario:
An online exam platform decides whether to warn, review, or block a student based on suspicious activity.
Detailed Description:
Read number of tab switches and camera status. If camera is off, print Blocked. If tab switches are 0, print Safe. If tab
switches are between 1 and 3, print Warning. If tab switches are more than 3, print Review Required.
Input Format:
An integer T and a string cameraStatus. cameraStatus is on or off.
Output Format:
Print Blocked, Safe, Warning, or Review Required.
Constraints:
0 <= T <= 100; cameraStatus will be on or off
Sample Input:
2 on
Sample Output:
Warning
Explanation:
Camera is on and tab switches are between 1 and 3, so the system gives a Warning.
'''
T, cameraStatus = input().split()
T = int(T)

if 0 <= T <= 100:
    if cameraStatus == "off":
        print("Blocked")
    elif T == 0:
        print("Safe")
    elif 1 <= T <= 3:
        print("Warning")
    else:
        print("Review Required")
else:
    print("Invalid")
```
