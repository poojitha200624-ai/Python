```python
'''
Question 1: Shopping Bill Total
Scenario
A small shop stores item prices in a list. Write a program to read the number of items and their prices, then print the
total bill amount.
Problem Statement
Write a Python program using list concepts to solve the above scenario.
Input Format
First line: integer n. Second line: n space-separated integers representing prices.
Output Format
Print the total price.
Constraints
1 <= n <= 100; 1 <= price <= 10000
Sample Input
5
10 20 30 40 50
Sample Output
150
Explanation
Add all elements of the list. 10 + 20 + 30 + 40 + 50 = 150
'''
n = int(input())
arr = list(map(int, input().split()))
print(sum(arr))
```

```python
'''
Question 2: Find the Highest Mark
Scenario
A teacher stores marks of students in a list. Write a program to find the highest mark.
Problem Statement
Write a Python program using list concepts to solve the above scenario.
Input Format
First line: integer n. Second line: n space-separated integers representing marks.
Output Format
Print the highest mark.
Constraints
1 <= n <= 100; 0 <= mark <= 100
'''
n = int(input())
arr=list(map(int,input().split()))
print(max(arr))
```

```python
'''
Question 3: Count Even Numbers
Scenario
A list contains daily token numbers. Write a program to count how many token numbers are even.
Problem Statement
Write a Python program using list concepts to solve the above scenario.
Input Format
First line: integer n. Second line: n space-separated integers.
Output Format
Print the count of even numbers.
Constraints
1 <= n <= 100; 0 <= number <= 10000
Sample Input
7
12 5 8 9 10 3 6
Sample Output
4
Explanation
The even numbers are 12, 8, 10, and 6, so the count is 4
'''
n = int(input())
arr = list(map(int,input().split()))
count = 0
for i in range(n):
  if arr[i] % 2 == 0:
    count+=1
print(count)
```

```python
'''
Question 4: Reverse the List
Scenario
A user enters a sequence of numbers. Write a program to print the list in reverse order.
Problem Statement
Write a Python program using list concepts to solve the above scenario.
Input Format
First line: integer n. Second line: n space-separated integers.
Output Format
Print the elements in reverse order separated by spaces.
Constraints
1 <= n <= 100
Sample Input
5
1 2 3 4 5
Sample Output
5 4 3 2 1
Explanation
Reverse the original list and print each element.
'''
n = int(input())
arr = list(map(int,input().split()))
print(*arr[::-1])
```

```python
'''
Question 5: Search an Element
Scenario
A list contains student roll numbers. Write a program to check whether a given roll number is present.
Problem Statement
Write a Python program using list concepts to solve the above scenario.
Input Format
First line: integer n. Second line: n space-separated roll numbers. Third line: integer target.
Output Format
Print FOUND if target is present, otherwise print NOT FOUND.
Constraints
1 <= n <= 100; 1 <= roll number <= 100000
Sample Input
Sample Output
5
101 102 103 104 105
103
Explanation
The target 103 exists in the list, so the output is FOUND.
'''
n = int(input())
arr = list(map(int, input().split()))
target = int(input())
if target in arr:
    print("FOUND")
else:
    print("NOT FOUND")
```

```python
'''
Question 6: Sum of Odd Numbers
Scenario
FOUND
A list contains numbers collected from a game. Write a program to find the sum of only odd numbers.
Problem Statement
Write a Python program using list concepts to solve the above scenario.
Input Format
First line: integer n. Second line: n space-separated integers.
Output Format
Print the sum of odd numbers.
Constraints
1 <= n <= 100; 0 <= number <= 10000
Sample Input
6
3 4 7 10 11 2
Sample Output
Explanation
Odd numbers are 3, 7, and 11. Their sum is 21
'''
n = int(input())
arr = list(map(int, input().split()))
total = 0
for num in arr:
    if num % 2 != 0:
        total += num
print(total)
```

```python
'''
Question 7: Remove Duplicates
Scenario
21
A list contains repeated product IDs. Write a program to print only unique IDs in the order they first appear.
Problem Statement
Write a Python program using list concepts to solve the above scenario.
Input Format
First line: integer n. Second line: n space-separated integers.
Output Format
Print unique elements separated by spaces.
Constraints
1 <= n <= 100; 1 <= ID <= 10000
Sample Input
8
4 2 4 5 2 7 5 9
Sample Output
4 2 5 7 9
Explanation
Keep an element only if it has not appeared before
'''
n = int(input())
arr = list(map(int, input().split()))
unique = []
for num in arr:
    if num not in unique:
        unique.append(num)
print(*unique)
```

```python
'''
Question 8: Second Largest Number
Scenario
A list contains scores. Write a program to find the second largest unique score.
Problem Statement
Write a Python program using list concepts to solve the above scenario.
Input Format
First line: integer n. Second line: n space-separated integers.
Output Format
Print the second largest unique number. If it does not exist, print NO SECOND LARGEST.
Constraints
1 <= n <= 100; -10000 <= number <= 10000
Sample Input
6
10 20 30 30 25 20
Sample Output
25
Explanation
Unique values are 10, 20, 25, and 30. The second largest is 25
'''
n = int(input())
arr = list(map(int, input().split()))
unique = []
for num in arr:
    if num not in unique:
        unique.append(num)
if len(unique) < 2:
    print("NO SECOND LARGEST")
else:
    unique.sort()
    print(unique[-2])
```

```python
'''
Question 9: List Rotation by One
Scenario
A queue list needs to be rotated to the right by one position. Write a program to move the last element to the first
position.
Problem Statement
Write a Python program using list concepts to solve the above scenario.
Input Format
First line: integer n. Second line: n space-separated integers.
Output Format
Print the rotated list separated by spaces.
Constraints
1 <= n <= 100
Sample Input
Sample Output
5
1 2 3 4 5
Explanation
5 1 2 3 4
The last element 5 moves to the front, and all other elements shift right
'''
n = int(input())
arr = list(map(int, input().split()))
last = arr.pop()
arr.insert(0, last)
print(*arr)
```

```python
'''
Question 10: Positive and Negative Count
Scenario
A list contains profit and loss values. Positive values are profit, negative values are loss, and zero is neutral. Count
positive and negative values.
Problem Statement
Write a Python program using list concepts to solve the above scenario.
Input Format
First line: integer n. Second line: n space-separated integers.
Output Format
Print two integers: positive_count negative_count.
Constraints
1 <= n <= 100; -10000 <= value <= 10000
Sample Input
7
10 -5 0 8 -2 -9 4
Sample Output
Explanation
Positive values are 10, 8, and 4. Negative values are -5, -2, and -9
'''
# Read input
n = int(input())
nums = list(map(int, input().split()))
positive_count = 0
negative_count = 0
# Count positive and negative numbers
for num in nums:
    if num > 0:
        positive_count += 1
    elif num < 0:
        negative_count += 1
# Print result
print(positive_count, negative_count)
```
