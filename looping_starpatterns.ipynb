```python
'''
1. ATM Withdrawal Feasibility Checker
Scenario
A bank ATM receives multiple withdrawal requests during a day. Each customer enters an amount, and the ATM must count
how many requests can be processed using only valid currency rules.
Detailed Description
Write a program to read n withdrawal amounts. For each amount, check whether it is positive and divisible by 100. Count
valid and invalid requests, and calculate the total valid withdrawal amount. Do not store the amounts; process each input
directly using loops.
Input Format
First line: integer n. Next n lines: one integer amount per request.
Output Format
Print the number of valid requests, invalid requests, and the total valid withdrawal amount.
Constraints
1 <= n <= 10^5, -10^9 <= amount <= 10^9
Sample Input
6
500
750
1200-100
0
300
Sample Output
Valid Requests: 3
Invalid Requests: 3
Total Amount: 2000
Explanation
500, 1200, and 300 are positive and divisible by 100. The remaining three values are invalid, so the total valid amount is
2000
'''
n = int(input())
valid_request = 0
invalid_request = 0
total_amount = 0
if 1 <= n <= 10**5:
    for i in range(n):
        amount = int(input())
        if -10**9 <= amount <= 10**9:
            if amount > 0 and amount % 100 == 0:
                valid_request += 1
                total_amount += amount

            else:
                invalid_request += 1
print("Valid Requests:", valid_request)
print("Invalid Requests:", invalid_request)
print("Total Amount:", total_amount)
```

```python
'''
2. Digital Root Without Formula
Scenario
A school ID verification system compresses a long numeric ID into a single digit checksum by repeatedly adding its digits.
Detailed Description
Read a positive integer and repeatedly calculate the sum of its digits until the result becomes a single digit. Use loops only;
do not use arrays, lists, strings, or direct digital-root formulas.
Input Format
One integer n.
Output Format
Print the final single-digit checksum.
Constraints
1 <= n <= 10^18
Sample Input
9875
Sample Output
2
Explanation
9 + 8 + 7 + 5 = 29, then 2 + 9 = 11, then 1 + 1 = 2. Therefore the checksum is 2
'''
n = int(input())
if 1 <= n <= 10**18:
  while n >= 10:
    s = 0
    while n > 0:
      digit = n % 10
      s += digit
      n // 10
    n = s
  print(n)
```

```python
'''
6. Prime Numbers in a Range Counter
Scenario
A math practice app asks students to count how many prime numbers exist between two values.
Detailed Description
Read two integers l and r. Count the prime numbers from l to r inclusive. Use nested loops for checking divisibility. Avoid
arrays, lists, and precomputed tables.
Input Format
Two integers l and r.
Output Format
Print the number of prime numbers in the range.
Constraints
1 <= l <= r <= 5000
Sample Input
10 20
Sample Output
Prime Count: 4
Explanation
The prime numbers from 10 to 20 are 11, 13, 17, and 19. Therefore the count is 4.'''
l, r = map(int, input().split())
count = 0
for num in range(l, r + 1):
    if num > 1:
        prime = True
        for i in range(2, num):
            if num % i == 0:
                prime = False
                break
        if prime:
            count += 1
print("Prime Count:", count)
```

```python
'''
7. Perfect Number Finder in Range
Scenario
A number theory trainer needs to identify how many perfect numbers are present in a small interval.
Detailed Description
Read l and r. A number is perfect if the sum of its proper divisors equals the number itself. Count perfect numbers in the
range using loops only.
Input Format
Two integers l and r.
Output Format
Print the count of perfect numbers in the given range.
Constraints
1 <= l <= r <= 10000
Sample Input
1 30
Sample Output
Perfect Number Count: 2
Explanation
6 and 28 are perfect numbers because their proper divisors add up to the same number
'''
l, r = map(int, input().split())
count = 0
for num in range(l, r + 1):
    divisor_sum = 0
    for i in range(1, num):
        if num % i == 0:
            divisor_sum += i
    if divisor_sum == num:
        count += 1
print("Perfect Number Count:", count)
```

```python
'''
8. Strong Number Checker
Scenario
A puzzle app checks whether a number is a strong number. A strong number is equal to the sum of factorials of its digits.
Detailed Description
Read a number n. For every digit, calculate its factorial using a loop and add it to a sum. Finally compare the sum with the
original number. Do not use arrays or precomputed factorial lists.
Input Format
One non-negative integer n.
Output Format
Print whether the number is a strong number.
Constraints
0 <= n <= 10^9
Sample Input
145
Sample Output
Strong Number: Yes
Explanation
1! + 4! + 5! = 1 + 24 + 120 = 145, so 145 is a strong number'''
n = int(input())
original = n
total = 0
while n > 0:
    digit = n % 10
    fact = 1
    for i in range(1, digit + 1):
        fact *= i
    total += fact
    n //= 10
if total == original:
    print("Strong Number: Yes")
else:
    print("Strong Number: No")
```

```python
'''
9. Armstrong Number Range Counter
Scenario
A coding contest asks participants to count all Armstrong numbers in a given range.
Detailed Description
Page 5
Read l and r. For each number, count its digits, then calculate the sum of each digit raised to that digit count. If the sum
equals the number, count it. Use loops only.
Input Format
Two integers l and r.
Output Format
Print the count of Armstrong numbers in the range.
Constraints
1 <= l <= r <= 100000
Sample Input
1 500
Sample Output
Armstrong Count: 13
Explanation
The range includes one-digit Armstrong numbers and numbers such as 153, 370, 371, and 407. The total count is 13.
'''
l, r = map(int, input().split())
count = 0
for num in range(l, r + 1):
    original = num
    # Count digits
    temp = num
    digits = 0
    while temp > 0:
        digits += 1
        temp //= 10
    # Special case for 0 (not needed here since l>=1, but safe)
    if original == 0:
        digits = 1
    # Find Armstrong sum
    temp = num
    total = 0
    while temp > 0:
        digit = temp % 10
        total += digit ** digits
        temp //= 10
    if total == original:
        count += 1
print("Armstrong Count:", count)
```

```python
'''
11. GCD Using Repeated Subtraction
Scenario
A basic calculator does not support modulo operation, so it finds the GCD using repeated subtraction.
Detailed Description
Read two positive integers a and b. Find their greatest common divisor by repeatedly subtracting the smaller value from the
larger value until both values become equal.
Input Format
Two positive integers a and b.
Output Format
Print the GCD.
Constraints
1 <= a,b <= 10^5
Sample Input
48 18
Sample Output
GCD: 6
Explanation
Repeated subtraction eventually makes both values equal to 6. Hence the GCD is 6
'''
a, b = map(int, input().split())

while a != b:
    if a > b:
        a = a - b
    else:
        b = b - a

print("GCD:", a)
```

```python
'''
12. LCM by Incremental Search
Scenario
A timetable system needs to find the first time two repeating events meet again.
Detailed Description
Page 7
Read two positive integers a and b. Find the least common multiple by starting from the larger number and increasing until a
value divisible by both numbers is found. Use loops only.
Input Format
Two positive integers a and b.
Output Format
Print the LCM.
Constraints
1 <= a,b <= 1000
Sample Input
12 18
Sample Output
LCM: 36
Explanation
36 is the smallest positive number divisible by both 12 and 18
'''
a, b = map(int, input().split())

lcm = max(a, b)

while True:
    if lcm % a == 0 and lcm % b == 0:
        break
    lcm += 1

print("LCM:", lcm)
```

```python
'''
14. Lucky Digit Frequency
Scenario
A game marks a number as lucky based on how many times a chosen digit appears in it.
Detailed Description
Read a non-negative number n and a digit d. Count how many times d appears in n using arithmetic digit extraction and
loops. Do not convert the number to a string.
Input Format
One non-negative integer n and one digit d.
Output Format
Print the frequency of digit d in n.
Constraints
0 <= n <= 10^18, 0 <= d <= 9
Sample Input
505050 5
Sample Output
Frequency: 3
Explanation
The digit 5 appears three times in 505050
'''
n, d = map(int, input().split())
count = 0
if n == 0 and d == 0:
    count = 1
else:
    while n > 0:
        digit = n % 10
        if digit == d:
            count += 1
        n //= 10
print("Frequency:", count)
```

```python
'''
15. Minimum Days to Reach Savings Goal
Scenario
A student saves money every day. The saved amount increases by a fixed value each day, and the student wants to know
when the goal is reached.
Detailed Description
Read starting daily saving s, daily increase inc, and target amount t. Using a loop, add each day savings to the total and
increase the next day saving by inc. Print the minimum number of days needed to reach or exceed the target.
Input Format
Three integers s, inc, and t.
Output Format
Print the minimum number of days required.
Constraints
1 <= s,inc,t <= 10^9
Sample Input
100 50 700
Sample Output
Days Required: 4
Explanation
Savings by day are 100, 150, 200, and 250. The total becomes 700 after 4 days, so the goal is reached in 4 days.
'''
s, inc, t = map(int, input().split())
total = 0
days = 0
daily = s
while total < t:
    total += daily
    days += 1
    daily += inc
print("Days Required:", days)
```
